```python
# 4 dersin not ortalamasını hesaplayıp harf notunu, en yüksek ve en düşük notu gösterir.
notlar = []
for i in range(1, 5):
    not_degeri = float(input(f"{i}. dersin notunu girin: "))
    notlar.append(not_degeri)

ort = sum(notlar) / len(notlar)

if ort >= 90:
    harf = "AA"
elif ort >= 80:
    harf = "BA"
elif ort >= 70:
    harf = "BB"
elif ort >= 60:
    harf = "CB"
elif ort >= 50:
    harf = "CC"
else:
    harf = "FF"

print(f"Ortalama: {ort} - Harf Notu: {harf}")
print("En yüksek not:", max(notlar))
print("En düşük not:", min(notlar))
```

raba Sınıfı
python
marka, model ve hiz (km/s) bilgilerini tutan bir Araba sınıfı oluşturun.
hızlandır(artis) metodu ile hızı artırın.
bilgi_goster() metodu ile araç bilgilerini yazdırın.
Dikdörtgen Sınıfı
python
uzunluk ve genislik bilgilerini tutan bir Dikdortgen sınıfı oluşturun.
alan_hesapla() metodu ile alanı hesaplayın (uzunluk * genislik).
cevre_hesapla() metodu ile çevreyi hesaplayın (2 * (uzunluk + genislik)).
Kitap Sınıfı
python
ad, yazar ve sayfa_sayisi bilgilerini tutan bir Kitap sınıfı oluşturun.
ozet() metodu ile "Suç ve Ceza - Dostoyevski (671 sayfa)" formatında bilgi döndürün.

```python
class Araba:
  def __init__(self,marka,model,hiz):
    self.marka = marka
    self.model = model
    self.hiz = hiz

  def hızlandir(self,artis):
    self.hiz += artis

  def bilgi_goster(self):
        print(f"Marka: {self.marka}, Model: {self.model}, Hız: {self.hiz} km/s")
```

```python
araba1 = Araba("Toyota", "Corolla", 120)
araba1.hızlandir(30)
araba1.bilgi_goster()
```

```python
class Dikdortgen:
  def __init__(self,uzunluk,genislik):
    self.uzunluk = uzunluk
    self.genislik = genislik

  def cevre_hesapla(self):
    return((self.uzunluk + self.genislik)*2)

  def alan_hesapla(self):
    return self.uzunluk * self.genislik
```

```python
Dikdortgen1 = Dikdortgen(3,5)
Dikdortgen1.cevre_hesapla()
Dikdortgen1.alan_hesapla()
```

```python
class Kitap:
  def __init__(self,ad,yazar,sayfa_sayisi):
    self.ad = ad
    self.yazar = yazar
    self.sayfa_sayisi = sayfa_sayisi

  def ozet(self):
    return f"{self.ad} - {self.yazar} ({self.sayfa_sayisi} sayfa)"


```

```python
kitap1 = Kitap("suc ve ceza","Dosteyevski","671")
kitap1.ozet()
```
