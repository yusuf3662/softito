```python
import pandas as pd

wine_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'

wine_df = pd.read_csv(wine_url, sep=';')

wine_df.head()
```

```python

X_wine_regression = wine_df.drop('quality', axis=1)
y_wine_regression = wine_df['quality']
```

```python
def classify_wine_quality(quality_score):
    if quality_score >= 8:
        return 'premium'
    elif quality_score >= 6:
        return 'good'
    elif quality_score >= 4:
        return 'average'
    elif quality_score == 3:
        return 'below average'
    else:
        return 'bad'

y_wine_classification = wine_df['quality'].apply(classify_wine_quality)
X_wine_classification = wine_df.drop('quality', axis=1)

pd.concat([X_wine_classification, y_wine_classification], axis=1).head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, accuracy_score, precision_score, recall_score, f1_score
import pandas as pd


X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_wine_regression, y_wine_regression, test_size=0.2, random_state=42
)

reg_model = LinearRegression()
reg_model.fit(X_train_reg, y_train_reg)
y_pred_reg = reg_model.predict(X_test_reg)

mae = mean_absolute_error(y_test_reg, y_pred_reg)
mse = mean_squared_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

regression_metrics = {
    'Metric': ['MAE', 'MSE', 'R-squared'],
    'Value': [mae, mse, r2]
}
regression_metrics_df = pd.DataFrame(regression_metrics)
```

```python

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_wine_classification, y_wine_classification, test_size=0.2, random_state=42, stratify=y_wine_classification
)

cls_model = RandomForestClassifier(random_state=42)
cls_model.fit(X_train_cls, y_train_cls)
y_pred_cls = cls_model.predict(X_test_cls)

accuracy = accuracy_score(y_test_cls, y_pred_cls)
precision = precision_score(y_test_cls, y_pred_cls, average='weighted', zero_division=0)
recall = recall_score(y_test_cls, y_pred_cls, average='weighted', zero_division=0)
f1 = f1_score(y_test_cls, y_pred_cls, average='weighted', zero_division=0)

classification_metrics = {
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-Score'],
    'Value': [accuracy, precision, recall, f1]
}
classification_metrics_df = pd.DataFrame(classification_metrics)



print("Regression Model Metrics")
display(regression_metrics_df)

print("Classification Model Metrics")
display(classification_metrics_df)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Logistic Regression model for classification
log_reg_model = LogisticRegression(max_iter=1000, random_state=42, solver='liblinear') # Increased max_iter for convergence
log_reg_model.fit(X_train_cls, y_train_cls)
y_pred_log_reg = log_reg_model.predict(X_test_cls)

# Calculate metrics for Logistic Regression
log_reg_accuracy = accuracy_score(y_test_cls, y_pred_log_reg)
log_reg_precision = precision_score(y_test_cls, y_pred_log_reg, average='weighted', zero_division=0)
log_reg_recall = recall_score(y_test_cls, y_pred_log_reg, average='weighted', zero_division=0)
log_reg_f1 = f1_score(y_test_cls, y_pred_log_reg, average='weighted', zero_division=0)

logistic_regression_metrics = {
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-Score'],
    'Value': [log_reg_accuracy, log_reg_precision, log_reg_recall, log_reg_f1]
}
logistic_regression_metrics_df = pd.DataFrame(logistic_regression_metrics)

print("Logistic Regression Model Metrikleri")
display(logistic_regression_metrics_df)
```

### TensorFlow Keras ile Sınıflandırma Modeli

Şimdi aynı sınıflandırma görevini bir TensorFlow Keras Sequential modeli kullanarak yeniden oluşturacağız. Bunun için önce veriyi ölçeklendirmemiz ve hedef değişkeni one-hot encode etmemiz gerekecek.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.preprocessing import StandardScaler, LabelEncoder
from tensorflow.keras.utils import to_categorical

scaler = StandardScaler()
X_train_cls_scaled = scaler.fit_transform(X_train_cls)
X_test_cls_scaled = scaler.transform(X_test_cls)


label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train_cls)
y_test_encoded = label_encoder.transform(y_test_cls)

y_train_categorical = to_categorical(y_train_encoded)
y_test_categorical = to_categorical(y_test_encoded)


num_classes = y_train_categorical.shape[1]

print(f"Eğitim veri setindeki sınıf sayısı: {num_classes}")
print(f"Eğitim özellikleri şekli: {X_train_cls_scaled.shape}")
print(f"Eğitim etiketleri şekli: {y_train_categorical.shape}")
```

```python

model = Sequential([
    Dense(128, activation='relu', input_shape=(X_train_cls_scaled.shape[1],)),
    Dense(64, activation='relu'),
    Dense(num_classes, activation='softmax')
])


model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])


history = model.fit(X_train_cls_scaled, y_train_categorical,
                    epochs=50,  # Deneme için epoch sayısını ayarlayın
                    batch_size=32,
                    validation_split=0.2,
                    verbose=0) # Eğitimin ilerlemesini görmek isterseniz 1 yapın

print("Keras modeli başarıyla eğitildi.")
```

```python

loss, accuracy = model.evaluate(X_test_cls_scaled, y_test_categorical, verbose=0)

```
